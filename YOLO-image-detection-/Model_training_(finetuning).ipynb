Fine-tuning means taking a pre-trained machine-learning model (one that has already learned general patterns from a large dataset) and then training it further on a smaller, task-specific dataset so that it performs better for your particular use case.


```python
from ultralytics import YOLO
from ultralytics import YOLO
# can use new updated models from this https://docs.ultralytics.com/models/yolo12/#usage-examples
# Load a COCO-pretrained YOLO12n model
model = YOLO("yolo12n.pt")

# Train the model on the COCO8 example dataset for 100 epochs # fine tune on your own data
results = model.train(data="coco8.yaml", epochs=100, imgsz=640)

```

```python
from ultralytics import YOLO

# Load your trained model (already trained weights)
model = YOLO("/content/yolo11n.pt")

# Predict on an image
results = model("https://img.freepik.com/premium-photo/portrait-dog-running_1048944-29290430.jpg?semt=ais_hybrid&w=740")  # replace with your image path

# Get class names from the result
names = results[0].names  # dictionary: class_id -> class_name

# Get detected class IDs
classes = results[0].boxes.cls.cpu().numpy().astype(int)

# Print unique detected class names
detected_classes = set([names[c] for c in classes])

for obj in detected_classes:
    print(obj)
```
