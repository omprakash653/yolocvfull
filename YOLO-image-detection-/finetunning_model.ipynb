```python
from ultralytics import YOLO

# Load a COCO-pretrained YOLOv8n model
model = YOLO("yolov8n.pt")

# Train the model on the COCO8 example dataset for 100 epochs # fine tune on our own data
results = model.train(data="coco8.yaml", epochs=100, imgsz=640)
```

```python
results.summary()
```

```python
from ultralytics import YOLO

# Load your trained model (if already trained)
model = YOLO("runs/train/exp/weights/yolo11n.pt")  # or use the updated model object if already trained

# Predict on an image
results = model("dogrunning.jpeg")  # replace with your image path

# Get class names from the result
names = results[0].names  # dictionary: class_id -> class_name
classes = results[0].boxes.cls.cpu().numpy().astype(int)  # detected class IDs

# Print unique detected class names
detected_classes = set([names[c] for c in classes])
for obj in detected_classes:
    print(obj)
```
